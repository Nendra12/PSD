# Business Understanding

## Latar Belakang Bisnis

Kondisi dan jenis permukaan jalan merupakan faktor penting dalam keselamatan dan kenyamanan berkendara. Jalan tanah, jalan berbatu (cobblestone), dan jalan beraspal memiliki karakteristik getaran yang berbeda ketika dilalui oleh kendaraan. Pemantauan kualitas dan jenis permukaan jalan masih banyak dilakukan secara manual melalui survei lapangan, yang memerlukan biaya dan waktu yang tidak sedikit. Di sisi lain, penggunaan smartphone yang telah dilengkapi sensor percepatan (accelerometer) membuka peluang untuk melakukan pemantauan kondisi jalan secara otomatis dan berkelanjutan. Oleh karena itu, dibutuhkan suatu sistem yang mampu mengklasifikasikan jenis permukaan jalan berdasarkan data accelerometer yang direkam oleh smartphone di dalam kendaraan.

## Permasalahan Bisnis

Permasalahan bisnis yang dihadapi adalah bagaimana melakukan identifikasi jenis permukaan jalan secara otomatis tanpa harus selalu mengirim petugas survei ke lapangan. Dinas terkait membutuhkan informasi mengenai bagian jalan mana yang berupa aspal, jalan tanah, ataupun jalan berbatu untuk keperluan perencanaan, pemeliharaan, maupun peningkatan infrastruktur. Oleh karena itu, diperlukan suatu pendekatan berbasis data untuk mengklasifikasikan jenis permukaan jalan dari sinyal percepatan yang dihasilkan oleh smartphone di dalam kendaraan.

## Tujuan Proyek

### Tujuan Bisnis

Tujuan bisnis dari proyek ini adalah menyediakan prototype sistem yang dapat membantu kepentingan, seperti pemerintah daerah atau pengelola infrastruktur jalan, dalam mengidentifikasi dan memetakan jenis permukaan jalan secara otomatis. Dengan adanya sistem ini, proses pemantauan kondisi jalan diharapkan menjadi lebih efisien, murah, dan dapat dilakukan secara lebih sering.

### Tujuan Data Mining

Tujuan data mining pada proyek ini adalah membangun dan mengevaluasi model klasifikasi time series yang mampu membedakan tiga jenis permukaan jalan, yaitu flexible pavement, cobblestone, dan dirt road, berdasarkan data univariat magnitudo percepatan yang direkam oleh smartphone. Model akan dikembangkan menggunakan algoritma klasifikasi tradisional (non–deep learning) dan dievaluasi menggunakan metrik seperti akurasi dan confusion matrix.

## Nilai Manfaat / Dampak

Jika model klasifikasi yang dibangun memiliki performa yang baik, sistem ini berpotensi digunakan sebagai bagian dari solusi pemantauan kondisi jalan secara skala besar. Data yang dikumpulkan dari banyak kendaraan dapat digunakan untuk membangun peta jenis permukaan jalan yang selalu diperbarui. Informasi tersebut bermanfaat bagi kepentingan yang bertanggung jawab dalam pengambilan keputusan, misalnya dalam menentukan prioritas perbaikan jalan, perencanaan rute transportasi, atau analisis keselamatan berkendara.

## Kriteria Keberhasilan

### Kriteria keberhasilan bisnis

Dari sisi bisnis, proyek ini dianggap berhasil apabila model yang dibangun mampu mengklasifikasikan jenis permukaan jalan dengan tingkat ketepatan yang memadai misal dengan akurasi yang “cukup baik” (≥ 75% atau 80%) , sehingga secara konsep dapat digunakan sebagai dasar sistem pemantauan kondisi jalan.

### Kriteria keberhasilan data mining

Dari sisi data mining, proyek ini dikatakan berhasil apabila model menghasilkan akurasi yang lebih baik dibandingkan baseline (misalnya prediksi kelas mayoritas), dan mampu membedakan ketiga kelas dengan cukup baik yang tercermin pada confusion matrix.

## Batasan & Asumsi

Proyek ini memiliki beberapa batasan. Pertama, data yang digunakan berasal dari kendaraan dan perangkat tertentu serta dikumpulkan pada jalan di beberapa kota di Brazil, sehingga karakteristiknya mungkin berbeda dengan kondisi jalan di Indonesia. Kedua, penelitian ini hanya menggunakan sinyal magnitudo accelerometer sebagai masukan model, tanpa mempertimbangkan informasi lain seperti GPS maupun kecepatan kendaraan. Ketiga, implementasi yang dilakukan bersifat offline dan terbatas pada pengujian model di lingkungan komputasi, belum sampai pada tahap integrasi dengan aplikasi seluler atau pengujian lapangan secara langsung.
