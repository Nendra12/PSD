# Business Understanding (No<sub>2</sub> Semarang)



## 1.1 Latar Belakang

Kualitas udara menjadi salah satu faktor penting dalam menjaga kesehatan masyarakat dan keberlanjutan lingkungan hidup. Salah satu indikator utama dalam pengukuran kualitas udara adalah konsentrasi **Nitrogen Dioksida (NO₂)**, gas polutan yang terutama berasal dari aktivitas pembakaran bahan bakar fosil seperti kendaraan bermotor, industri, dan pembangkit listrik.

Konsentrasi NO₂ yang tinggi dapat menyebabkan berbagai dampak negatif, mulai dari gangguan pernapasan, iritasi paru-paru, hingga memperburuk penyakit asma dan kardiovaskular. Selain itu, NO₂ juga berperan dalam pembentukan ozon troposfer dan hujan asam yang merusak lingkungan.

Oleh karena itu, kemampuan untuk **memprediksi konsentrasi NO₂ di masa mendatang** menjadi penting agar pemerintah, lembaga lingkungan, dan masyarakat dapat mengambil langkah pencegahan yang tepat sebelum kualitas udara memburuk.  

---

## 1.2 Permasalahan Bisnis

Hingga saat ini, sebagian besar pemantauan kualitas udara masih dilakukan secara **reaktif**, yaitu setelah pencemaran terjadi. Padahal, kemampuan untuk melakukan **prediksi kadar NO₂ secara harian** akan sangat membantu dalam pengambilan keputusan yang lebih proaktif.  

Namun, tantangan yang dihadapi adalah:
- Bagaimana memanfaatkan data historis konsentrasi NO₂ untuk memprediksi kadar NO₂ pada hari berikutnya?  
- Bagaimana memilih pendekatan model yang sederhana namun efektif untuk menghasilkan prediksi yang akurat dan mudah diimplementasikan?

---

## 1.3 Tujuan Bisnis

Proyek ini bertujuan untuk **membangun sistem prediksi konsentrasi NO₂ harian** menggunakan pendekatan *machine learning*, khususnya algoritma **K-Nearest Neighbors Regression (KNN Regression)**.  

Adapun tujuan utamanya meliputi:
1. Membuat model prediksi kadar NO₂ berdasarkan data historis dari hari-hari sebelumnya.  
2. Menghasilkan estimasi konsentrasi NO₂ harian yang dapat digunakan untuk memberikan peringatan dini terhadap potensi pencemaran udara.  
3. Menyediakan dasar analisis yang dapat membantu pihak terkait dalam memantau tren kualitas udara.  

---

## 1.4 Manfaat yang Diharapkan

- **Bagi Pemerintah / Lembaga Lingkungan:**  
  Mendukung proses pemantauan kualitas udara dengan sistem prediksi sederhana dan hemat sumber daya.  

- **Bagi Masyarakat:**  
  Memberikan informasi prediktif mengenai kondisi udara pada hari berikutnya untuk membantu merencanakan aktivitas luar ruangan dengan lebih aman.  

- **Bagi Peneliti / Akademisi:**  
  Menjadi contoh penerapan model regresi non-parametrik (KNN Regression) dalam bidang prediksi lingkungan berbasis data terbuka.  

---

## 1.5 Ruang Lingkup Proyek

Proyek ini difokuskan pada **prediksi konsentrasi NO₂ harian** di wilayah tertentu berdasarkan data yang diperoleh dari **Copernicus Data Space Ecosystem (melalui API openEO)**.  

Ciri utama dataset:
- Data diambil dari citra satelit *Copernicus Sentinel* (produk pemantauan atmosfer).  
- Variabel utama: **konsentrasi NO₂ per hari**.  
- Periode data: beberapa hari atau bulan terakhir (historical daily data).  
- Data digunakan untuk membangun model prediksi kadar NO₂ pada **hari berikutnya (t+1)**.  

Pendekatan model yang digunakan adalah **K-Nearest Neighbors Regression (KNN Regression)**, yang memprediksi nilai berdasarkan kesamaan data historis (jarak antara titik data).  

Model ini dipilih karena:
- Sederhana dan mudah diterapkan.  
- Tidak mengasumsikan bentuk hubungan linear antar data.  
- Cocok untuk data lingkungan yang sering bersifat non-linear dan fluktuatif.  

---

## 1.6 Indikator Keberhasilan

1. **Indikator Teknis (Model Performance):**
   - Nilai error prediksi (RMSE, MAE, MAPE) berada pada tingkat yang rendah.  
   - Model mampu menangkap tren perubahan kadar NO₂ dengan stabilitas yang baik.  

2. **Indikator Bisnis (Manfaat Praktis):**
   - Prediksi kadar NO₂ dapat digunakan untuk memberikan **peringatan dini** terhadap potensi polusi.  
   - Meningkatkan kesiapsiagaan pemerintah atau masyarakat dalam menghadapi kondisi udara buruk.
