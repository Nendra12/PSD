```python
from scipy.io import arff
import pandas as pd
```

```python
data, meta = arff.loadarff("AsphaltPavementType_eq_TRAIN.arff")
df = pd.DataFrame(data)

# 3. Kadang label/kategori masih bertipe bytes, ubah ke string
for col in df.columns:
    if df[col].dtype == object:   # biasanya kolom class/label
        df[col] = df[col].astype(str)

# 4. Simpan ke CSV
df.to_csv("AsphaltPavementType_TRAIN.csv", index=False)

print("Selesai! File CSV tersimpan sebagai AsphaltPavementType.csv")
```
