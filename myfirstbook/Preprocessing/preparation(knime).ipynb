# Preprocessing Data (Ecoli) by Knime

Ini adalah tahapan preprocessing data Missing Value, Outlier Detection, dan Balancing Data menggunakan aplikasi Knime Analytics 

## Data 

Berikut data ecoli yang sudah di import ke database postgresql

![Ecoli PostgreSQL](../image/postgresql_ecoli.jpg)


## Koneksi Database

Buka aplikasi Knime kemudian cari node bernama **PostgreSQL Connector** kemudian drag and drop ke bagian field di kanan

![PostgreSQL Connector](../image/knime.gif)

setelah itu klik dua kali bagian node nya dan setting hostname, database, username, password dll

![PostgreSQL Connector Setting](../image/postgresql_connector.jpg)

setelah di setting execute node dengan klik kanan node dan klik "execute"

![PostgreSQL Connector Execute](../image/postgresql_connector_execute.jpg)

## Memilih Table 

Setelah connect ke database PostgreSQL pilih table yang ingin digunakan contoh di khasus ini menggunakan table ecoli dengan menambahkan node **DB Selector** dan koneksikan node *PostgreSQL Connector* ke *DB Selector* dengan kabel

![PostgreSQL Connector](../image/db_selector.jpg)

Kemudian setting 

![PostgreSQL Connector](../image/db_selector_setting.jpg)

dan execute

## Membaca Data

Tambahkan node *DB Reader* untuk membaca data di table

![PostgreSQL Connector](../image/db_reader.jpg)

dan execute

## Melihat Missing Value

Untuk menangani missing value kita perlu mengetahui apakah ada missing value atau tidak dengan menggunakan node **Statistics** dan visualkan menggunakan node *Bar Chart*

![PostgreSQL Connector](../image/statistik.jpg)

![PostgreSQL Connector](../image/missing_value_bar.png)

grafik menunjukan tidak ditemukannya **Missing Value**, jadi pada tahap selanjutnya tambahkan node missing value (Opsional) *Wajib* jika grafik menunjukan adanya missing value.

![PostgreSQL Connector](../image/missing_value.jpg)

## Outlier Detection

Untuk mengecek dan menangani adanya outlier menggunakan node **Numeric Outliers** dan *Bar Chart* untuk visualisasi

![PostgreSQL Connector](../image/outlier_add.jpg)

klik dua kali pada node dan setting seperti ini

![PostgreSQL Connector](../image/outlier.jpg)

kemudian execute

![PostgreSQL Connector](../image/outlier_bar.png)

Berdasarakan grafik terdapat outlier di beberapa fitur dan akan di tangani dengan Raplace Outlier Values *bisa anda setting pada settingan node
