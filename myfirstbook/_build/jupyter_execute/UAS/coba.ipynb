```python
import pandas as pd

df = pd.read_csv("Datasets/AsphaltPavementType_combine.csv")
df2 = pd.read_csv("Datasets/AsphaltPavementType_TEST.csv")

subsetA = df.iloc[1055:]

# pastikan kolom yang dibandingkan sama
common_cols = df2.columns.intersection(subsetA.columns).tolist()

# buang duplikat supaya set lebih kecil
subsetA_unique = subsetA[common_cols].drop_duplicates()

# bikin "kunci" per baris (string) agar mudah dicek
keyA = subsetA_unique.astype(str).agg("|".join, axis=1)

key2 = df2[common_cols].astype(str).agg("|".join, axis=1)

# hapus dari df2 yang ada di subsetA
df2_clean = df2[~key2.isin(set(keyA))].reset_index(drop=True)

print("df2 sebelum:", df2.shape)
print("df2 sesudah:", df2_clean.shape)

df2_clean.to_csv("Datasets/AsphaltPavementType_TEST_New.csv")
```
