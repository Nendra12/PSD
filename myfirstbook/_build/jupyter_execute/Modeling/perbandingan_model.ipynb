# Perbandingan Akurasi Model (ECOLI)

| Kondisi Data                   | Naive Bayes | Random Forest | Bagging (Naive Bayes) | Bagging (Decision Tree) |
|--------------------------------|-------------|---------------|------------------------|--------------------------|
| Sebelum Oversampling            | 0.720       | 0.852         | 0.750                  | **0.867**                |
| Sesudah Oversampling **SMOTE**  | 0.825       | **0.969**     | 0.829                  | 0.965                    |
| Sesudah Oversampling **ADASYN** | 0.857       | **0.939**     | 0.857                  | 0.935                    |

## Kesimpulan
1. Semua model mengalami peningkatan akurasi setelah dilakukan oversampling, baik menggunakan SMOTE maupun ADASYN.  
2. Model **Random Forest** konsisten memberikan akurasi tertinggi pada data hasil oversampling, baik SMOTE (0.969) maupun ADASYN (0.939).  
3. Sebelum oversampling, **Bagging dengan Decision Tree** memberikan performa terbaik (0.867).  
4. Secara keseluruhan, **Random Forest** adalah model paling unggul untuk dataset ECOLI setelah dilakukan oversampling.
