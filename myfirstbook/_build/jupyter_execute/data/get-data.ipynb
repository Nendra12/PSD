# Get Data (iris)

Halaman ini menjelaskan cara menghubungkan Python ke **MySQL** dan **PostgreSQL** menggunakan **SQLAlchemy** dan **pandas**, agar proses pengambilan data menjadi lebih ringkas dan efisien.

---

## Persiapan Awal

Sebelum memulai, pastikan sudah menginstal package berikut:

```bash
pip install sqlalchemy pymysql psycopg2 pandas
```

Keterangan:  
- **SQLAlchemy** → library utama untuk membuat koneksi database.  
- **pymysql** → driver untuk MySQL.  
- **psycopg2** → driver untuk PostgreSQL.  
- **pandas** → membaca hasil query langsung ke dalam DataFrame.

---

## Koneksi ke MySQL

Berikut contoh koneksi dan pengambilan data dari MySQL menggunakan SQLAlchemy:

```python
import pandas as pd
from sqlalchemy import create_engine

# Buat koneksi ke MySQL
engine = create_engine("mysql+pymysql://root:@localhost/iris")

# Ambil data dari tabel
df = pd.read_sql("SELECT * FROM iris", engine)

# Tampilkan data
print(df.head())

# Tutup koneksi
engine.dispose()
```

**Penjelasan singkat:**
- Format koneksi:  
  ```
  mysql+pymysql://<user>:<password>@<host>/<nama_database>
  ```
- `pandas.read_sql()` membaca hasil query SQL langsung ke DataFrame.
- `engine.dispose()` digunakan untuk menutup koneksi setelah selesai.

---

## Koneksi ke PostgreSQL

Contoh koneksi PostgreSQL dengan format SQLAlchemy hampir sama, hanya berbeda pada driver dan port default.

```python
import pandas as pd
from sqlalchemy import create_engine

# Buat koneksi ke PostgreSQL
engine = create_engine("postgresql+psycopg2://postgres:@localhost:5432/iris")

# Ambil data dari tabel
df = pd.read_sql("SELECT * FROM iris", engine)

# Tampilkan data
print(df.head())

# Tutup koneksi
engine.dispose()
```

**Format koneksi PostgreSQL:**
```
postgresql+psycopg2://<user>:<password>@<host>:<port>/<nama_database>
```
